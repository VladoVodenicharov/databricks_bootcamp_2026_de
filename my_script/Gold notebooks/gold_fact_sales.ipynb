# Gold layer notebook to extract fact sales data and load it into the gold schema of the catalog. This is the STAR schema fact table

```python
query = """
SELECT
    sd.order_id,
    pr.product_key,
    cu.customer_key,
    sd.order_date,
    sd.ship_date,
    sd.due_date,
    sd.sales AS sales_amount,
    sd.quantity,
    sd.price
FROM workspace.silver.crm_sales sd
LEFT JOIN workspace.silver.crm_product cp
    ON sd.product_id = cp.product_key
LEFT JOIN workspace.gold.dim_products pr
    ON cp.product_id = pr.product_id
LEFT JOIN workspace.gold.dim_customers cu
    ON sd.customer_id = cu.customer_id
"""
df = spark.sql(query)
```

## Write the query as a delta table in the gold catalog

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.gold.fact_sales")
```
